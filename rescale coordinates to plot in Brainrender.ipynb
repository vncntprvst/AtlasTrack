```python
# import the coordinates from cell_count. Change the format to DataFrame
import pandas as pd
oo5 = pd.read_csv(‘/Users/jun/Documents/Project/Orofacial premotor circuits in the adult - RV tracing and Manipulation/Whisking/2018_0115 007retro-4 RV tracing/Serial section images - processed/count - fixed 20200408/roi_table_007retro_wp4_20200408.txt’)
```

```python
# Pick AP, DV and ML columns
file_names = [‘oo5’,‘oo7’,‘mas2’]
for i in range(0,len(file_names)):
    file_names[i] = eval(file_names[i])[[‘AP_location’ ,‘DV_location’, ‘ML_location’]]
```

```python
# Change column names to x, y, z.
for i in range(0,len(file_names)):
    file_names[i] = file_names[i].rename(columns={‘AP_location’: ‘x’, ‘DV_location’ :‘y’, ‘ML_location’:‘z’})
```

```python
# rescale
for i in range(0,len(file_names)):
    file_names[i].x = -file_names[i].x*1000 + 5400
    file_names[i].y = file_names[i].y*1000;
    file_names[i].z = file_names[i].z*1000 +5700;
```
